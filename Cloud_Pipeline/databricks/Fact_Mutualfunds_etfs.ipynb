```python
s3_path = "s3://travel-analytics-bronze/mutualfund_etfs/Fact_Mutualfunds_etfs/Fact_Mutualfunds_etfs.csv"

df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(s3_path)
)

df.display()
```

```python
spark.sql("""
CREATE DATABASE IF NOT EXISTS finance_analytics
""")
```

```python
df.write \
  .mode("overwrite") \
  .saveAsTable("finance_analytics.fact_mutualfunds_etfs")
```
